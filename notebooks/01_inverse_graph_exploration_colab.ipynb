# Exploratory inverse Collatz graph

This notebook is an executable research record for a finite inverse expansion. The forward map is $T(n)=n/2$ for even $n$ and $T(n)=3n+1$ for odd $n$. We store edges in the forward direction, while the construction discovers nodes by traversing predecessor branches.

The experiment is deliberately finite: `max_nodes` controls computational truncation. Results must therefore be reported together with the root set, budget, software versions, and plotting seed.

```python
import logging

import matplotlib.pyplot as plt
import pandas as pd

from collatz_graph import InverseGraphConfig, build_inverse_graph
from collatz_graph.analysis import graph_node_table, graph_summary
from collatz_graph.plotting import plot_inverse_graph

logging.basicConfig(level=logging.INFO, format='%(levelname)s:%(name)s:%(message)s')
config = InverseGraphConfig(roots=(1,), max_nodes=5_000, show_progress=True)
graph = build_inverse_graph(config)
summary = graph_summary(graph)
summary
```

```python
node_table = graph_node_table(graph)
display(node_table.head(10))
display(node_table.groupby('inverse_depth', as_index=False).agg(nodes=('node', 'size'), max_node=('node', 'max')))
```

```python
plot_inverse_graph(graph, node_size=12)
plt.show()
```

## Interpretation and experimental risks

The node table describes only the induced finite graph produced by the configured breadth-first expansion. A maximum depth, degree distribution, or node magnitude is censored by the node budget and should not be interpreted as a property of the infinite inverse graph.

Potential sources of error include accidental changes to the root order, budget-dependent frontier truncation, software-version differences in layout algorithms, and treating spring-layout coordinates as mathematical structure. The next useful extension is to export the configuration and summary alongside each experiment, then repeat the analysis across several budgets and roots.
